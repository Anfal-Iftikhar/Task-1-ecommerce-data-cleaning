**Import Libraries**

```python
import pandas as pd
import numpy as np
```

**Load Dataset**

```python
# Load dataset
df = pd.read_excel('/Dataset for Data Analytics.xlsx')
```

**Explore Dataset**

```python
# Show first 5 rows
df.head()
```

**CHECK DATASET INFORMATION**

```python
# Dataset information
df.info()
```

**CHECK TOTAL ROWS & COLUMNS**

```python
print("Rows and Columns:", df.shape)
```

**CHECK COLUMN NAMES**

```python
print(df.columns)
```

**CHECK MISSING VALUES**

```python
# Check missing values
df.isnull().sum()
```

**HANDLE MISSING VALUES**

```python
# Replace missing CouponCode values
df['CouponCode'] = df['CouponCode'].fillna('No Coupon')
```

**VERIFY MISSING VALUES AGAIN**

```python
df.isnull().sum()
```

**CHECK DUPLICATE ROWS**

```python
# Check duplicate rows
duplicates = df.duplicated().sum()
print("Duplicate Rows:", duplicates)
```

**REMOVE DUPLICATES (IF ANY)**

```python
# Remove duplicate rows
df = df.drop_duplicates()
```

**CHECK DUPLICATE ORDER IDs**

```python
# Check duplicate OrderIDs
duplicate_ids = df['OrderID'].duplicated().sum()
print("Duplicate OrderIDs:", duplicate_ids)
```

**CHECK DATE FORMAT**

```python
# Convert Date column to proper datetime format
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
```

**CHECK INVALID DATES**

```python
# Check invalid dates
invalid_dates = df['Date'].isnull().sum()

print("Invalid Dates:", invalid_dates)
```

**FORMAT DATE COLUMN**

```python
# Standardize date format
df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')
```

**CHECK DATA TYPES**

```python
print(df.dtypes)
```

**CLEAN TEXT COLUMNS**

```python
# Remove extra spaces from text columns
text_columns = [
    'Product',
    'ShippingAddress',
    'PaymentMethod',
    'OrderStatus',
    'CouponCode',
    'ReferralSource'
]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()
```

**STANDARDIZE TEXT FORMAT**

```python
# Capitalize text properly
df['PaymentMethod'] = df['PaymentMethod'].str.title()
df['OrderStatus'] = df['OrderStatus'].str.title()
df['ReferralSource'] = df['ReferralSource'].str.title()
```

**VALIDATE TOTAL PRICE**

```python
# Create validation column
df['PriceCheck'] = np.where(
    df['Quantity'] * df['UnitPrice'] == df['TotalPrice'],
    'Correct',
    'Mismatch'
)
```

**FINAL QUALITY CHECK**

```python
print("FINAL PROJECT CHECK")

print("---------------------------")

print("Missing Values:")
print(df.isnull().sum())

print("---------------------------")

print("Duplicate Rows:", df.duplicated().sum())

print("---------------------------")

print("Duplicate OrderIDs:",
      df['OrderID'].duplicated().sum())

print("---------------------------")

print("Invalid Dates:",
      df['Date'].isnull().sum())
```
